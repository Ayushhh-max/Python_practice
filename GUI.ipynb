```python
#Combobox Widget
import tkinter as tk
from tkinter.ttk import Combobox
window = tk.Tk()
window.title("Combobox Example")
window.geometry("800x200")
def combobox_selected(event):
    selected_value = combobox.get()
    label.config(text=f"Selected: {selected_value}")
label = tk.Label(window, text="Select an option:")
label.pack(padx=20, pady=10)
options = ["Option 1", "Option 2", "Option 3"]
combobox = Combobox(window, values=options)
combobox.pack(padx=20, pady=10)
combobox.bind("<<ComboboxSelected>>",
              combobox_selected)
window.mainloop()
```

```python
#Login Form Using Grid
from tkinter import *
window = Tk()
window.title("Login Example")
def submit_clicked():
    result_label.config(text="Good Job")
Label(window, text="Username:").grid(row=0, column=0, padx=10, pady=5)
Entry(window).grid(row=0, column=1, padx=10, pady=5)
Label(window, text="Password:").grid(row=1, column=0, padx=10, pady=5)
Entry(window, show="*").grid(row=1, column=1, padx=10, pady=5)
Button(window,
       text="Submit",command=submit_clicked).grid(row=2,column=0,
       columnspan=2,pady=20)
result_label = Label(window, text="")
result_label.grid(row=3,column=0,columnspan=2)
window.mainloop()
```

```python
#Sum Calculator
from tkinter import *
window = Tk()
window.title("Sum Calculator")
window.geometry("800x200")
def calculate_sum():
    num1 = float(entry1.get())
    num2 = float(entry2.get())
    total = num1 + num2
    result_label.config(
        text=f"Sum: {total}"
    )
Label(window, text="Number 1:").grid(
    row=0, column=0)
entry1 = Entry(window)
entry1.grid(row=0, column=1)
Label(window, text="Number 2:").grid(
    row=1, column=0)
entry2 = Entry(window)
entry2.grid(row=1, column=1)
Button(window,
       text="Calculate Sum",
       command=calculate_sum).grid(
       row=2,
       column=0,
       columnspan=2)
result_label = Label(window, text="Sum:")
result_label.grid(
    row=3,
    column=0,
    columnspan=2)
window.mainloop()
```

```python
#Arithmetic Calculator Project
from tkinter import *

window = Tk()
window.title("Arithmetic Calculator")
window.geometry("800x200")

def calculate():
    try:
        num1 = float(entry1.get())
        num2 = float(entry2.get())
    except ValueError:
        result_label.config(
            text="Please enter valid numbers."
        )
        return

    operation = operation_var.get()

    if operation == "+":
        result = num1 + num2
    elif operation == "-":
        result = num1 - num2
    elif operation == "*":
        result = num1 * num2
    elif operation == "/":
        if num2 == 0:
            result_label.config(
                text="Cannot divide by zero."
            )
            return
        result = num1 / num2
    else:
        result_label.config(
            text="Please select an operation."
        )
        return

    result_label.config(
        text=f"Result: {result}"
    )

Label(window, text="Number 1:").grid(
    row=0, column=0)

entry1 = Entry(window)
entry1.grid(row=0, column=1)

Label(window, text="Number 2:").grid(
    row=1, column=0)

entry2 = Entry(window)
entry2.grid(row=1, column=1)

operation_var = StringVar()
operation_var.set("+")

frame = Frame(window)
frame.grid(row=2, column=1)

Radiobutton(frame, text="+",
            variable=operation_var,
            value="+").pack(side=LEFT)

Radiobutton(frame, text="-",
            variable=operation_var,
            value="-").pack(side=LEFT)

Radiobutton(frame, text="*",
            variable=operation_var,
            value="*").pack(side=LEFT)

Radiobutton(frame, text="/",
            variable=operation_var,
            value="/").pack(side=LEFT)

Button(window,
       text="Calculate",
       command=calculate).grid(
       row=3,
       column=0,
       columnspan=2)

result_label = Label(window,
                     text="Result:")
result_label.grid(
    row=4,
    column=0,
    columnspan=2)

window.mainloop()
```
